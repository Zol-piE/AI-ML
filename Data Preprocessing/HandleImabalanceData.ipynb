##Setup

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
from sklearn.model_selection import train_test_split
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 50)
```

##Load Dataset

```python
import kagglehub
import shutil
path = kagglehub.dataset_download("tomalislam/diabetes-dataset-for-practice")
print("Path to dataset files:", path)
shutil.copytree(path, '/content/dataset', dirs_exist_ok=True, ignore=shutil.ignore_patterns('.nfs*'))
```

```python
df = pd.read_csv('/content/dataset/diabetes.csv')
```

```python
df.head()
```

##Split Dataset

```python
target = 'Outcome'
X = df.drop(target,axis=1)
y = df[target]
```

```python
xTrain,xTest,yTrain,yTest = train_test_split(X,y,test_size=.2,random_state=42,stratify=y)
```

```python
print(xTest.shape)
print(yTest.shape)
print(xTrain.shape)
print(yTrain.shape)
```

##Target Distribution

```python
print(yTrain.value_counts())
print(yTrain.value_counts(normalize=True)*100)
print('-'*50)
print(yTest.value_counts())
print(yTest.value_counts(normalize=True)*100)
print('-'*50)
```

```python
from imblearn.under_sampling import RandomUnderSampler
from imblearn.over_sampling import  RandomOverSampler , SMOTE
```

##Random Under Sample (Train Only)

```python
sampler = RandomUnderSampler(random_state=42)
```

```python
xTrainResample , yTrainResample = sampler.fit_resample(xTrain,yTrain)
```

```python
print(yTrain.value_counts())
print(yTrain.value_counts(normalize=True)*100)
print('-'*50)
print(yTrainResample.value_counts())
print(yTrainResample.value_counts(normalize=True)*100)
print('-'*50)
```

##Random Over Sample

```python
sampler = RandomOverSampler(random_state=42)
xTrainResample,yTrainResample = sampler.fit_resample(xTrain,yTrain)
```

```python
print(yTrain.value_counts())
print(yTrain.value_counts(normalize=True)*100)
print('-'*50)
print(yTrainResample.value_counts())
print(yTrainResample.value_counts(normalize=True)*100)
print('-'*50)
```

##SMOTE

```python
sampler = SMOTE(random_state=42)
xTrainResample , yTrainResample = sampler.fit_resample(xTrain,yTrain)
print(yTrain.value_counts())
print(yTrain.value_counts(normalize=True)*100)
print('-'*50)
print(yTrainResample.value_counts())
print(yTrainResample.value_counts(normalize=True)*100)
print('-'*50)
```
