##Setup

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
from sklearn.model_selection import train_test_split
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 50)
```

##Load Dataset

```python
import kagglehub
import shutil
path = kagglehub.dataset_download("tomalislam/titanic-dataset-fot-practice")
shutil.copytree(path,'/content/Dataset',dirs_exist_ok=True)
df = pd.read_csv('/content/Dataset/titanic_dataset.csv')
df.head()
```

```python
print(df.shape)
print(df.columns.to_list())
```

###Drop rows with missing target

```python
print(df.isnull().sum())
```

```python
target = 'survived'
missingValuesTarget = df[target].isna().sum()
print(missingValuesTarget)
df = df.dropna(subset=[target])
print(df.shape)
```

###Separate X(feature) and y(target)

```python
X = df.drop(target, axis=1)
y = df[target]
```

```python
X[:5]
```

```python
y[:5]
```

```python
print('Target Distribution on full data :')
print(y.value_counts())
print('-'*50)
print("In percentage ")
print(y.value_counts(normalize=True)*100)
```

###Train Test Split


```python
xTrain , xTest, yTrain,yTest = train_test_split(X,y,test_size=.2,random_state=42,stratify=y)
print(xTrain.shape)
print(xTest.shape)
print(yTrain.shape)
print(yTest.shape)
```

```python
#straitify distributes like target y
print(yTrain.value_counts(normalize=True)*100)
print(yTest.value_counts(normalize=True)*100)
```

##Save Data

```python
dfTrain = xTrain.copy()
dfTrain[target] = yTrain.values
dfTest = xTest.copy()
dfTest[target] = yTest.values
```

```python
dfTrain.to_csv('train.csv', index=False)
dfTest.to_csv('test.csv', index=False)
```
