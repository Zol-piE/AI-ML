#Multivariate Analysis
* pairplot
* correlation heatmap
* Cluster Map
* interaction and subgroup pattern discovery

##Setup

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

##Load dataset

```python
df = pd.read_csv('/content/titanic_dataset.csv')
```

```python
df['survivedLabel'] = df['survived'].map({0:'no',1:'yes'})
df.head()
```

```python
numCol = ['age','fare','sibsp','parch']
catCol = ['survived', 'pclass', 'sex', 'embarked', 'class', 'who', 'adult_male', 'deck', 'embark_town', 'alone']
```

##Pairplot

```python
dfPair = df[numCol + ['survivedLabel']]
plt.figure(figsize=(10,10))
sns.pairplot(dfPair,hue='survivedLabel')
```

##Scatterplot

```python
plt.figure(figsize=(5,3))
sns.scatterplot(data=df, x='age', y='fare', hue='survivedLabel', style='sex', alpha=.7)
plt.show()
```

##Correlation Heatmap

```python
corr = df[numCol].corr()
print(corr)
```

```python
plt.figure(figsize=(5,3))
sns.heatmap(corr,annot=True,cmap='coolwarm')
plt.show()
```

##Cluster Map

```python
plt.figure(figsize=(3,3))
sns.clustermap(corr,annot=True,cmap='coolwarm')
```

##Interaction and Subgroup Pattern

```python
plt.figure(figsize=(6,4))
sns.scatterplot(x=df['age'], y=df['fare'], hue=df['survivedLabel'])
plt.show()
```

###Subgroups: sibilings/spouses

```python
plt.figure(figsize=(5,3))
sns.scatterplot(x=df['age'],y=df['fare'],hue=df['sibsp'],palette='viridis')
plt.show()
```
