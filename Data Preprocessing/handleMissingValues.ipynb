##Setup

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
from sklearn.model_selection import train_test_split
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 50)
```

##Load Dataset

```python
import kagglehub
import shutil
path = kagglehub.dataset_download("tomalislam/titanic-dataset-fot-practice")
shutil.copytree(path,'/content/Dataset',dirs_exist_ok=True)
df = pd.read_csv('/content/Dataset/titanic_dataset.csv')
dfTest = pd.read_csv('/content/Dataset/test.csv')
dfTrain = pd.read_csv('/content/Dataset/train.csv')
```

##Split data

```python
target = 'survived'
xTrain = dfTrain.drop(target,axis=1)
xTest = dfTest.drop(target,axis=1)
yTrain = dfTrain[target]
yTest = dfTest[target]
```

```python
print(f'Test Shape :{dfTest.shape}')
print(f'Train Shape :{dfTrain.shape}')
print(f'feature train shape : {xTrain.shape}')
print(f'output train shape :  {yTrain.shape}')
print(f'feature test shape : {xTest.shape}')
print(f'output test shape :  {yTest.shape}')
```

##Find Missing Values

```python
missingValueCount = xTrain.isnull().sum()
missingValuepercentage = xTrain.isna().mean() *100
missing = pd.DataFrame({
    'missingValueCount' : missingValueCount,
    'missingValuePercentage' : missingValuepercentage
})
missing.sort_values(by='missingValueCount',ascending=False)
```

##Found Columns with High missing Values and Drop

```python
threshold = 40
highMissingCol = missing[missing['missingValuePercentage'] > threshold].index.to_list()
print(highMissingCol)
```

```python
xTrain.drop(highMissingCol,axis=1,inplace=True)
xTest.drop(highMissingCol,axis=1,inplace=True)
```

```python
print(xTrain.shape)
print(xTest.shape)
```

##Filter Columns

```python
numCol = ["age", "sibsp", "parch", "fare"]
catCol = ['pclass', 'sex', 'embarked', 'class', 'who', 'adult_male', 'embark_town', 'alone']

print("Numerical columns:", numCol)
print("Categorical columns:", catCol)
```

##Imputaion
>Imputation should be done to both X_train and X_test, but the imputation values must be calculated using X_train only.

###Numerical Data Imputaion

```python
xTrainImputed = xTrain.copy(deep=True)
xTestImputed = xTest.copy(deep=True)

for col in numCol:
    median = xTrainImputed[col].median()

    xTrainImputed[col] = xTrainImputed[col].fillna(median)
    xTestImputed[col] = xTestImputed[col].fillna(median)
```

###Categorical Data Imputaion

```python
for col in catCol:
    mode = xTrain[col].mode()[0]
    print(f"Column->'{col}' mode value ->{mode}")
    xTrainImputed[col] = xTrainImputed[col].fillna(mode)
    xTestImputed[col] = xTestImputed[col].fillna(mode)
```

##Save Imputed data

```python
imputedTrain = xTrainImputed.copy(deep=True)
imputedTrain[target] = yTrain
imputedTrain.to_csv('imputedTrain.csv',index=False)

imputedTest = xTestImputed.copy(deep=True)
imputedTest[target] = yTest
imputedTest.to_csv('imputedTest.csv',index=False)
```

```python
missingValueCount = imputedTrain.isnull().sum()
missingValuepercentage = imputedTrain.isna().mean() *100
missing = pd.DataFrame({
    'missingValueCount' : missingValueCount,
    'missingValuePercentage' : missingValuepercentage
})
missing.sort_values(by='missingValueCount',ascending=False)
```
