> **Standardization** expresses a value in terms of how many standard deviations it is from the mean.

> **Normalization** expresses a value's position relative to the minimum and maximum of the feature.

##Setup

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, MinMaxScaler
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 50)
```

##Load Dataset

```python
import kagglehub
import shutil
path = kagglehub.dataset_download("tomalislam/diabetes-dataset-for-practice")
print("Path to dataset files:", path)
shutil.copytree(path, '/content/dataset', dirs_exist_ok=True, ignore=shutil.ignore_patterns('.nfs*'))
```

```python
df = pd.read_csv('/content/dataset/diabetes.csv')
```

##Split Dataset

```python
target = 'Outcome'
X = df.drop(target,axis=1)
y = df[target]
```

```python
xTrain,xTest,yTrain,yTest = train_test_split(X,y,test_size=.2,random_state=42,stratify=y)
```

```python
print(xTest.shape)
print(yTest.shape)
print(xTrain.shape)
print(yTrain.shape)
```

```python
numCol = xTrain.select_dtypes([np.number]).columns.to_list()
catCol = xTrain.columns.difference(numCol).to_list()
print(numCol)
print(catCol)
```

##Standardization
> mean ~ 0 , standard deviation ~ 1

> calculated using z-score = x - mean / std

```python
scaler = StandardScaler()
xTestScaled = scaler.fit_transform(xTest[numCol])
xTrainScaled = scaler.transform(xTrain[numCol])
```

```python
print(xTrain[numCol].describe().T)
print('-'*50)
xTrainScaled = pd.DataFrame(xTrainScaled,columns=numCol)
print(xTrainScaled.describe().T)
```
